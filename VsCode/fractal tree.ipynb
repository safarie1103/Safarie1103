```python
import turtle

def draw_fractal_tree(branch_len, t, angle, level):
    if level == 0:
        return
    t.forward(branch_len)
    t.left(angle)
    draw_fractal_tree(branch_len * 0.7, t, angle, level - 1)
    t.right(2 * angle)
    draw_fractal_tree(branch_len * 0.7, t, angle, level - 1)
    t.left(angle)
    t.backward(branch_len)

screen = turtle.Screen()
t = turtle.Turtle()
t.speed(0)
t.left(90)
draw_fractal_tree(100, t, 30, 5)
screen.mainloop()
```
